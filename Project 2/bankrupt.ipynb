```python
import pandas as pd
import matplotlib.pyplot as plt

# --- Étape 1 : définir les blocs et leurs DataFrames ---
blocs = {
    "Profit": df_profit,
    "EPS": df_eps,
    "CashFlow": df_cashflow,
    "Growth": df_growth,
    "Liquidity": df_liquidity,
    "Efficiency": df_efficiency
}

# --- Étape 2 : récupérer la variable la plus corrélée avec Bankrupt pour chaque bloc ---
summary = {}

for bloc_name, df_bloc in blocs.items():
    corr = df_bloc.corr()["Bankrupt"].drop("Bankrupt")  # corrélation avec Bankrupt
    best_var = corr.abs().idxmax()                        # colonne avec |corr| max
    best_value = corr.loc[best_var]                       # valeur réelle avec signe
    summary[bloc_name] = best_value

# --- Étape 3 : convertir en Series et trier pour le graphe ---
summary_series = pd.Series(summary).sort_values()
print(summary_series)

# --- Étape 4 : tracer le bar plot ---
plt.figure(figsize=(8,5))
summary_series.plot(kind="barh", color="skyblue")
plt.title("Impact des variables clés par bloc sur la faillite")
plt.xlabel("Corrélation avec Bankrupt")
plt.ylabel("Bloc")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()

#Le graphique révèle que la faillite est avant tout une affaire de performance financière : la rentabilité (Profit) tire la sonnette d’alarme, mais la liquidité joue un rôle encore plus protecteur que les bénéfices par action (EPS) ou le cash-flow. L’efficacité opérationnelle pèse un peu, et la croissance brille par son absence d’influence. En clair, mieux vaut gérer son argent et sa dette que simplement viser la croissance pour éviter la faillite.
```

```python
import pandas as pd

# Suppose que df est ton DataFrame principal avec Bankrupt et toutes les colonnes renommées
# On définit d'abord les colonnes clés
profit_col = "ROA_A"
eps_col = "NetVal_A"
cashflow_col = "CashFlowPerShare"
efficiency_col = "OpProfitPerPerson"
debt_col = "DebtRatioPct"

# --- 1. Moyennes par faillite ---
kpi_means = df.groupby("Bankrupt")[[profit_col, eps_col, cashflow_col]].mean()
kpi_means.rename(index={1:"Faillite", 2:"Non-faillite"}, inplace=True)

# --- 2. Moyenne Debt Ratio ---
debt_mean = df.groupby("Bankrupt")[debt_col].mean()

# --- 3. Moyenne Efficiency ---
efficiency_mean = df.groupby("Bankrupt")[efficiency_col].mean()

# --- 4. % d'entreprises à risque élevé ---
# Critères : Profit bas (ROA_A < median) et DebtRatio haut (> median)
profit_median = df[profit_col].median()
debt_median   = df[debt_col].median()

high_risk = df[(df[profit_col] < profit_median) & (df[debt_col] > debt_median)]
high_risk_pct = len(high_risk) / len(df) * 100

# --- Affichage synthétique ---
print("=== Moyennes Profit, EPS, CashFlow par Bankrupt ===")
print(kpi_means)
print("\n=== Moyenne Debt Ratio par Bankrupt ===")
print(debt_mean)
print("\n=== Moyenne Efficiency par Bankrupt ===")
print(efficiency_mean)
print(f"\n% d'entreprises à risque élevé (Profit bas + DebtRatio haut) : {high_risk_pct:.2f}%")
```
