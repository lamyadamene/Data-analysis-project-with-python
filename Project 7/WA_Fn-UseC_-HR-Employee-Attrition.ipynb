```python
df =df[["Attrition",
    "Age",
    "Gender",
    "MaritalStatus",
    "Department",
    "JobRole",
    "DistanceFromHome",
    "MonthlyIncome",
    "PercentSalaryHike",
    "JobSatisfaction",
    "WorkLifeBalance",
    "YearsAtCompany"]]
```

```python
print(df['YearsAtCompany'].max())
```

```python
df['Attrition_bin']= df['Attrition'].map({'Yes': 1, 'No': 0})
```

```python
Attrition_rate = df['Attrition_bin'].mean()
print(f'The attrition rate is {Attrition_rate:.2%}')
```

```python
attrition_dept_gender = (
    df.groupby(["Department", "Gender"])["Attrition_bin"]
    .mean()
    .unstack()
)

attrition_dept_gender.plot(
    kind="barh",
    title="Attrition rate par département et sexe"
)
```

```python
attrition_by_role = (
    df.groupby(["JobRole", "Gender"])["Attrition_bin"]
    .mean()
    .unstack()
)

attrition_by_role.plot(kind="barh", title="Attrition rate par métier")
```

```python
# Moyenne d'attrition par JobSatisfaction et Gender
attrition_job_gender = (
    df.groupby(["JobSatisfaction", "Gender"])["Attrition_bin"]
    .mean()
    .unstack()
)

# Graphe horizontal
attrition_job_gender.plot(
    kind="barh",
    title="Attrition selon JobSatisfaction et sexe"
)
```

```python
attrition_wlb_gender = (
    df.groupby(["WorkLifeBalance", "Gender"])["Attrition_bin"]
    .mean()
    .unstack()
)

attrition_wlb_gender.plot(
    kind="barh",
    title="Attrition selon WorkLifeBalance et sexe"
)
```

```python
df["Distance_bin"] = pd.cut(
    df["DistanceFromHome"],
    bins=[0, 5, 10, 20, 30],
    labels=["0-5 km", "6-10 km", "11-20 km", "21-30 km"]
)

# Attrition par Distance + Gender
attrition_dist_gender = (
    df.groupby(["Distance_bin", "Gender"])["Attrition_bin"]
    .mean()
    .unstack()
)

# Graphe horizontal
attrition_dist_gender.plot(
    kind="barh",
    title="Attrition selon distance domicile-travail et sexe"
)
```

```python
import streamlit as st
import plotly.express as px
import pandas as pd

st.set_page_config(
    page_title="Dashboard Carte Bancaire",
    layout="wide"
)

st.title("📊 Dashboard Carte Bancaire")

# Exemple de données
df = pd.DataFrame({
    "Statut": ["OK", "Refus", "En attente"],
    "Nombre": [1200, 300, 150]
})

fig = px.bar(df, x="Statut", y="Nombre")

# Layout en colonnes (VRAI dashboard)
col1, col2 = st.columns(2)

with col1:
    st.plotly_chart(fig, use_container_width=True)

with col2:
    st.plotly_chart(fig, use_container_width=True)
```
